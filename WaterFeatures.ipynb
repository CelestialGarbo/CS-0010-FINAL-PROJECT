```python
import pandas as pd
df = pd.read_csv("PittWaterFeatures.csv")
df.head()

neighbor_count = {}
for index, row in df.iterrows():
    n = row["neighborhood"]
    if(n in neighbor_count):
        neighbor_count[n] += 1
    else:
        neighbor_count[n] = 1
print(neighbor_count)

import matplotlib.pyplot as plt

plt.bar(neighbor_count.keys(), neighbor_count.values())
plt.xticks(rotation=90)
plt.show()

```
