```python
import pandas as pd
df = pd.read_csv("PittWaterFeatures.csv")
df.head()

neighbor_count = {}
for index, row in df.iterrows():
    n = row["neighborhood"]
    if(n in neighbor_count):
        neighbor_count[n] += 1
    else:
        neighbor_count[n] = 1
print(neighbor_count)

import matplotlib.pyplot as plt
print()
print("neighborhoods versus corresponding water feature count")
print()
plt.bar(neighbor_count.keys(), neighbor_count.values())
plt.xticks(rotation=90)
plt.show()

water_count = {}
for index, row in df.iterrows():
    n= row["feature_type"]
    if(n in water_count):
        water_count[n] += 1
    else:
        water_count[n] = 1
print(water_count)
print()
drinking_count = {}
for index, row in df.iterrows():
    n = row["neighborhood"]
    d = row["feature_type"]
    if(d == "Drinking Fountain" and n in drinking_count):
        drinking_count[n] += 1
    elif(d == "Drinking Fountain"):
        drinking_count[n] = 1
print(drinking_count)
print()
print("neighborhoods versus number of drinking fountains")
print()
plt.bar(drinking_count.keys(), drinking_count.values())
plt.xticks(rotation=90)
plt.show()


dec_count = {}
for index, row in df.iterrows():
    n = row["neighborhood"]
    d = row["feature_type"]
    if(d == "Decorative" and n in dec_count):
        dec_count[n] += 1
    elif(d == "Decorative"):
        dec_count[n] = 1
print(dec_count)
print()
print("neighborhoods versus number of Decorative features")
print()
plt.bar(dec_count.keys(), dec_count.values())
plt.xticks(rotation=90)
plt.show()

print()
sp_count = {}
for index, row in df.iterrows():
    n = row["neighborhood"]
    d = row["feature_type"]
    if(d == "Spray" and n in sp_count):
        sp_count[n] += 1
    elif(d == "Spray"):
        sp_count[n] = 1
print(sp_count)
print()
print("neighborhoods versus number of Spray features")
print()
plt.bar(sp_count.keys(), sp_count.values())
plt.xticks(rotation=90)
plt.show()
print()
p_count = {}
for index, row in df.iterrows():
    n = row["neighborhood"]
    d = row["feature_type"]
    if(d == "Pond" and n in p_count):
        p_count[n] += 1
    elif(d == "Pond"):
        p_count[n] = 1
print(p_count)
print()
print("neighborhoods versus number of Ponds")
print()
plt.bar(p_count.keys(), p_count.values())
plt.xticks(rotation=90)
plt.show()

```
