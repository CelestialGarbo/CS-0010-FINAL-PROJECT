```python
import pandas as pd
df = pd.read_csv("PittWaterFeatures.csv")
df.head()

import matplotlib.pyplot as plt
water_count = {}
for index, row in df.iterrows():
    n= row["feature_type"]
    if(n in water_count):
        water_count[n] += 1
    else:
        water_count[n] = 1
print("These are the counts for types of water features in all of Pittsburgh")        
print(water_count)
print("First the data will be assessed for every type of feature with a sorted dictionary and corresponding graph")
```

```python
import pandas as pd
df = pd.read_csv("PittWaterFeatures.csv")
df.head()

neighbor_count = {}
for index, row in df.iterrows():
    n = row["neighborhood"]
    if(n in neighbor_count):
        neighbor_count[n] += 1
    else:
        neighbor_count[n] = 1
print("Dictionary for nieghborhoods and total water feature counts")
print()
sorted_neighbors = dict(sorted(neighbor_count.items(), key = lambda item: item[1], reverse = True))

print(sorted_neighbors)


import matplotlib.pyplot as plt
print()

plt.figure(figsize = (14, 6))
plt.bar(sorted_neighbors.keys(), sorted_neighbors.values())
plt.xticks(rotation=90)
plt.xlabel("Neighborhoods")
plt.ylabel("Number of Water Features")
plt.title("Water Features per Neighborhood 2026")
plt.show()


```

```python
import pandas as pd
df = pd.read_csv("PittWaterFeatures.csv")
df.head()
import matplotlib.pyplot as plt

drinking_count = {}
for index, row in df.iterrows():
    n = row["neighborhood"]
    d = row["feature_type"]
    if(d == "Drinking Fountain" and n in drinking_count):
        drinking_count[n] += 1
    elif(d == "Drinking Fountain"):
        drinking_count[n] = 1

sorted_drinking = dict(sorted(drinking_count.items(), key = lambda item: item[1], reverse = True))
print("Dictionary of Drinking Fountains in Pittsburgh Neighborhoods")
print(sorted_drinking)
print()
plt.figure(figsize = (14, 6))
plt.bar(sorted_drinking.keys(), sorted_drinking.values())
plt.xticks(rotation=90)
plt.xlabel("Neighborhoods of Pittsburgh")
plt.ylabel("Number of Drinking Fountains")
plt.title("Drinking Fountains per Neighborhood 2026")
plt.show()
print()
print("These are the rankings for the best overall water features for the neighborhoods")
print()

keys_list = list(sorted_drinking.keys())
values_list = list(sorted_drinking.values())

ranking = {}
count = 1
for items in keys_list:
    ranking[items] = count
    count += 1
print(ranking)

print()
```
