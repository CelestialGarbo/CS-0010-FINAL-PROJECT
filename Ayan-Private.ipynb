This notebook will explore the frequency of fire incidents over time in different neighborhoods found around Pittsburgh. Fire safety is a very important subject when it comes to determining how safe a neighborhood is to live in, and this can be deemed a very important to look at to understand what the best/safest neighborhood in Pittsburgh is.

# 

```python
import pandas as pd
import matplotlib.pyplot as plt

# Load the data (it's tab-separated)
df = pd.read_csv("Fire-Incidents.tsv", sep="\t")

# Turn the alarm_time column into real dates
df["alarm_time"] = pd.to_datetime(df["alarm_time"])

# Keep only incidents from 2025
df_2025 = df[df["alarm_time"].dt.year == 2025]

# Count how many incidents happened in each neighborhood
counts = df_2025["neighborhood"].value_counts()

# Make the bar graph
counts.plot(kind="bar", figsize=(14, 6))
plt.title("Fire Incidents by Neighborhood (2025)")
plt.xlabel("Neighborhood")
plt.ylabel("Number of Fire Incidents")
plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# Load the data (it's tab-separated)
df = pd.read_csv("Fire-Incidents.tsv", sep="\t")

# Count how many incidents happened in each neighborhood
counts = df["neighborhood"].value_counts()

# Make the bar graph
counts.plot(kind="bar", figsize=(14, 6))
plt.title("Fire Incidents by Neighborhood (All Time)")
plt.xlabel("Neighborhood")
plt.ylabel("Number of Fire Incidents")
plt.tight_layout()
plt.show()
```

The graph above displays the frequency of reported fire incidents for every neighborhood in Pittsburgh of all the recorded data in the list. From the organization method used in earlier, the graph clearly indicates that the most fire-safe neighborhood in Pittsburgh is Mount Oliver Borough.
